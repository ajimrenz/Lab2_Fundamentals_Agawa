```python
# ============================================================
# ADDITIONAL EXERCISE 1
# SENSOR READING VALIDATOR
# ============================================================

import random

# ------------------------------------------------------------
# 1. CONFIGURATION
# ------------------------------------------------------------

SURNAME = "Agawa"
SEED_NUM = 2

print("=" * 50)
print("SENSOR READING VALIDATOR")
print("=" * 50)

print(f"Operator: {SURNAME}")
print(f"Seed Number: {SEED_NUM}")


# ------------------------------------------------------------
# 2. GENERATE A FIXED NUMBER OF SENSOR READINGS
# ------------------------------------------------------------

random.seed(SEED_NUM)

NUMBER_OF_READINGS = 10

sensor_data = []

for i in range(NUMBER_OF_READINGS):
    reading = random.uniform(0, 100)
    sensor_data.append(round(reading, 2))

print("\nGenerated Sensor Data:")
print(sensor_data)


# ------------------------------------------------------------
# 3. VALIDATE EACH SENSOR READING
# ------------------------------------------------------------

valid_results = []

print("\nValidation Results:")

for i, reading in enumerate(sensor_data, start=1):

    if 0 <= reading <= 100:
        result = "Valid"
    else:
        result = "Invalid"

    valid_results.append(result)

    print(f"Reading {i}: {reading} -> {result}")


# ------------------------------------------------------------
# 4. CLASSIFY VALID SENSOR READINGS
# ------------------------------------------------------------

classification_results = []

print("\nClassification Results:")

for i, reading in enumerate(sensor_data, start=1):

    if reading < 30:
        classification = "Low"

    elif reading < 70:
        classification = "Normal"

    else:
        classification = "High"

    classification_results.append(classification)

    print(f"Reading {i}: {reading} -> {classification}")


# ------------------------------------------------------------
# 5. SUMMARY
# ------------------------------------------------------------

valid_count = valid_results.count("Valid")
invalid_count = valid_results.count("Invalid")

low_count = classification_results.count("Low")
normal_count = classification_results.count("Normal")
high_count = classification_results.count("High")


print("\n" + "=" * 50)
print("FINAL SENSOR MONITORING SUMMARY")
print("=" * 50)

print(f"Operator: {SURNAME}")
print(f"Seed Number: {SEED_NUM}")
print(f"Total Readings: {NUMBER_OF_READINGS}")

print("\nValidation Summary:")
print(f"Valid Readings: {valid_count}")
print(f"Invalid Readings: {invalid_count}")

print("\nClassification Summary:")
print(f"Low: {low_count}")
print(f"Normal: {normal_count}")
print(f"High: {high_count}")

print("=" * 50)


# ------------------------------------------------------------
# 6. EXECUTION LOG
# ------------------------------------------------------------

print("\nExecution Log:")
print("1. Sensor readings generated.")
print("2. Each reading was validated.")
print("3. Valid readings were classified.")
print("4. Summary of results was displayed.")
print("5. Sensor validation process completed.")
```

```python
# ============================================================
# ADDITIONAL EXERCISE 2
# SIGNAL CHARACTER ANALYSIS
# ============================================================

# ------------------------------------------------------------
# 1. CONFIGURATION
# ------------------------------------------------------------

SURNAME = "Agawa"
SEED_NUM = 2

print("=" * 55)
print("SIGNAL CHARACTER ANALYSIS")
print("=" * 55)

print(f"Surname: {SURNAME}")
print(f"SEED_NUM: {SEED_NUM}")


# ------------------------------------------------------------
# 2. GENERATE THE SIGNAL
# ------------------------------------------------------------

generated_signal = SURNAME + str(SEED_NUM)

print("\nGenerated Signal:")
print(generated_signal)


# ------------------------------------------------------------
# 3. PROCESS / NORMALIZE THE SIGNAL
# ------------------------------------------------------------

processed_signal = generated_signal.strip().upper()

print("\nProcessed Signal:")
print(processed_signal)


# ------------------------------------------------------------
# 4. ANALYZE THE SIGNAL CHARACTER BY CHARACTER
# ------------------------------------------------------------

letter_count = 0
number_count = 0
space_count = 0
other_count = 0

character_analysis = []

for character in processed_signal:

    if character.isalpha():
        character_type = "Letter"
        letter_count += 1

    elif character.isdigit():
        character_type = "Number"
        number_count += 1

    elif character.isspace():
        character_type = "Space"
        space_count += 1

    else:
        character_type = "Other Character"
        other_count += 1

    character_analysis.append(
        f"'{character}' -> {character_type}"
    )


# ------------------------------------------------------------
# 5. DISPLAY CHARACTER ANALYSIS
# ------------------------------------------------------------

print("\nCharacter Analysis:")
print("-" * 55)

for analysis in character_analysis:
    print(analysis)


# ------------------------------------------------------------
# 6. DISPLAY CHARACTER COUNTS
# ------------------------------------------------------------

print("\nCharacter Summary:")
print("-" * 55)

print(f"Letters: {letter_count}")
print(f"Numbers: {number_count}")
print(f"Spaces: {space_count}")
print(f"Other Characters: {other_count}")


# ------------------------------------------------------------
# 7. FINAL CLASSIFICATION
# ------------------------------------------------------------

if other_count > 0:
    signal_classification = "Contains Other Characters"

elif number_count > 0 and letter_count > 0:
    signal_classification = "Alphanumeric Signal"

elif letter_count > 0:
    signal_classification = "Alphabetic Signal"

elif number_count > 0:
    signal_classification = "Numeric Signal"

elif space_count > 0:
    signal_classification = "Space-Based Signal"

else:
    signal_classification = "Empty Signal"


# ------------------------------------------------------------
# 8. FINAL OUTPUT
# ------------------------------------------------------------

print("\n" + "=" * 55)
print("FINAL OUTPUT")
print("=" * 55)

print(f"Generated Signal: {generated_signal}")
print(f"Processed Signal: {processed_signal}")
print(f"Signal Classification: {signal_classification}")


# ------------------------------------------------------------
# 9. EXECUTION LOG
# ------------------------------------------------------------

print("\nExecution Log:")
print("-" * 55)

print("1. Signal generated using surname and SEED_NUM.")
print("2. Signal was normalized.")
print("3. Signal was examined character by character.")
print("4. Characters were classified.")
print("5. Final signal classification was generated.")

print("=" * 55)
```

```python
# ============================================================
# ADDITIONAL EXERCISE 3
# PASSWORD AUTHENTICATION SYSTEM
# ============================================================

# ------------------------------------------------------------
# 1. CONFIGURATION
# ------------------------------------------------------------

SURNAME = "Agawa"
SEED_NUM = 2

print("=" * 55)
print("PASSWORD AUTHENTICATION SYSTEM")
print("=" * 55)

print(f"Operator: {SURNAME}")
print(f"SEED_NUM: {SEED_NUM}")


# ------------------------------------------------------------
# 2. GENERATE UNIQUE PASSWORD
# ------------------------------------------------------------

generated_password = SURNAME.upper() + str(SEED_NUM)

print("\nSystem initialized.")
print("Authentication required.")


# ------------------------------------------------------------
# 3. SET ATTEMPT LIMIT
# ------------------------------------------------------------

ATTEMPT_LIMIT = 3
attempts_made = 0
access_granted = False

execution_log = []


# ------------------------------------------------------------
# 4. AUTHENTICATION PROCESS
# ------------------------------------------------------------

while attempts_made < ATTEMPT_LIMIT:

    password = input("\nEnter password: ")

    attempts_made += 1

    if password == generated_password:

        access_granted = True

        execution_log.append(
            f"Attempt {attempts_made}: Authorized"
        )

        print("\nAccess Granted.")
        print("Authentication successful.")

        break

    else:

        remaining = ATTEMPT_LIMIT - attempts_made

        execution_log.append(
            f"Attempt {attempts_made}: Unauthorized"
        )

        print("\nAccess Denied.")

        if remaining > 0:
            print(f"Remaining attempts: {remaining}")

        else:
            print("Attempt limit reached.")
            print("System terminated.")


# ------------------------------------------------------------
# 5. FINAL SYSTEM STATE
# ------------------------------------------------------------

print("\n" + "=" * 55)
print("FINAL SYSTEM STATE")
print("=" * 55)

if access_granted:
    system_state = "UNLOCKED"
else:
    system_state = "LOCKED"

print(f"Generated Password: {generated_password}")
print(f"Attempt Limit: {ATTEMPT_LIMIT}")
print(f"Attempts Made: {attempts_made}")
print(f"Access Result: {'Granted' if access_granted else 'Denied'}")
print(f"Final System State: {system_state}")


# ------------------------------------------------------------
# 6. EXECUTION LOG
# ------------------------------------------------------------

print("\nExecution Log:")
print("-" * 55)

for log in execution_log:
    print(log)

print("=" * 55)
```
